```python
DATA_PATH = "/kaggle/input/datasets/ellipticco/elliptic-data-set/elliptic_bitcoin_dataset/"
```

```python
FEATURES_PATH = DATA_PATH + "elliptic_txs_features.csv"
CLASSES_PATH = DATA_PATH + "elliptic_txs_classes.csv"
EDGES_PATH = DATA_PATH + "elliptic_txs_edgelist.csv"
```

```python
features = pd.read_csv(
    FEATURES_PATH,
    header=None
)
```

```python
classes = pd.read_csv(CLASSES_PATH)
```

```python
edges = pd.read_csv(EDGES_PATH)
```

```python
print("Features:", features.shape)
print("Classes:", classes.shape)
print("Edges:", edges.shape)
```

```python
edges = pd.read_csv(EDGES_PATH)
```

```python
feature_names = ["txId", "Time_step"]

for i in range(1, 166):
    feature_names.append(f"feature_{i}")

features.columns = feature_names
```

```python
print("===== MISSING VALUES =====")

print("Features:", features.isnull().sum().sum())
print("Classes:", classes.isnull().sum().sum())
print("Edges:", edges.isnull().sum().sum())


print("\n===== DUPLICATES =====")

print("Duplicate feature rows:", features.duplicated().sum())
print("Duplicate txIds:", features["txId"].duplicated().sum())

print("Duplicate class rows:", classes.duplicated().sum())
print("Duplicate class txIds:", classes["txId"].duplicated().sum())

print("Duplicate edges:", edges.duplicated().sum())


print("\n===== CLASS DISTRIBUTION =====")

print(classes["class"].value_counts())
print("\nPercentage:")
print(classes["class"].value_counts(normalize=True) * 100)


print("\n===== TIME STEPS =====")

print("Number of time steps:", features["Time_step"].nunique())
print("Time steps:", sorted(features["Time_step"].unique()))


print("\n===== EDGE VALIDATION =====")

feature_txids = set(features["txId"])

source_missing = (~edges["txId1"].isin(feature_txids)).sum()
target_missing = (~edges["txId2"].isin(feature_txids)).sum()

print("Sources not in features:", source_missing)
print("Targets not in features:", target_missing)


print("\n===== CLASS/FEATURE ID VALIDATION =====")

feature_ids = set(features["txId"])
class_ids = set(classes["txId"])

print("Feature transactions:", len(feature_ids))
print("Class transactions:", len(class_ids))

print("Feature IDs missing from classes:", len(feature_ids - class_ids))
print("Class IDs missing from features:", len(class_ids - feature_ids))


print("\n===== CONSTANT FEATURES =====")

constant_features = []

for col in features.columns:
    if col not in ["txId", "Time_step"]:
        if features[col].nunique() <= 1:
            constant_features.append(col)

print("Number of constant features:", len(constant_features))
print("Constant features:", constant_features)
```

```python
feature_stats = features.describe().T

display(feature_stats)
```

```python
feature_stats[['mean', 'std', 'min', '25%', '50%', '75%', 'max']]
```

```python
numeric_features = features.drop(columns=["txId", "Time_step"])

feature_stats = numeric_features.describe().T

display(feature_stats)
```

```python
extreme_counts = pd.DataFrame({
    "above_5": (numeric_features.abs() > 5).sum(),
    "above_10": (numeric_features.abs() > 10).sum(),
    "above_50": (numeric_features.abs() > 50).sum(),
    "above_100": (numeric_features.abs() > 100).sum()
})

display(
    extreme_counts[
        (extreme_counts > 0).any(axis=1)
    ].sort_values("above_5", ascending=False)
)
```

```python
print("Unique classes:")
print(classes["class"].unique())

print("\nClass data type:")
print(classes["class"].dtype)

print("\nClass counts:")
print(classes["class"].value_counts())
```

```python
print("Number of nodes:", features["txId"].nunique())
print("Number of edges:", len(edges))

print("\nUnique source nodes:", edges["txId1"].nunique())
print("Unique destination nodes:", edges["txId2"].nunique())

print("\nAverage edges per node:", len(edges) / features["txId"].nunique())
```

```python
in_degree = edges.groupby("txId2").size()
out_degree = edges.groupby("txId1").size()

print("IN-DEGREE")
print(in_degree.describe())

print("\nOUT-DEGREE")
print(out_degree.describe())

print("\nMaximum in-degree:", in_degree.max())
print("Maximum out-degree:", out_degree.max())
```

```python
all_txids = features["txId"]

in_degree_full = all_txids.map(in_degree).fillna(0)
out_degree_full = all_txids.map(out_degree).fillna(0)

print("Nodes with zero in-degree:", (in_degree_full == 0).sum())
print("Nodes with zero out-degree:", (out_degree_full == 0).sum())

print("\nNodes with both zero:", ((in_degree_full == 0) & (out_degree_full == 0)).sum())

print("\nTotal degree statistics:")
total_degree = in_degree_full + out_degree_full
print(total_degree.describe())
```

```python
node_features = features.copy()

node_features["in_degree"] = in_degree_full
node_features["out_degree"] = out_degree_full
node_features["total_degree"] = (
    node_features["in_degree"] +
    node_features["out_degree"]
)

print(node_features.shape)
display(
    node_features[
        ["txId", "Time_step", "in_degree", "out_degree", "total_degree"]
    ].head(10)
)
```

```python
node_features["fan_in_out_ratio"] = (
    node_features["in_degree"] /
    node_features["out_degree"].replace(0, 1)
)

print(node_features.shape)

display(
    node_features[
        [
            "txId",
            "in_degree",
            "out_degree",
            "total_degree",
            "fan_in_out_ratio"
        ]
    ].head(10)
)
```

```python
unique_in_neighbors = edges.groupby("txId2")["txId1"].nunique()
unique_out_neighbors = edges.groupby("txId1")["txId2"].nunique()

unique_in_neighbors_full = (
    all_txids.map(unique_in_neighbors).fillna(0)
)

unique_out_neighbors_full = (
    all_txids.map(unique_out_neighbors).fillna(0)
)

print("Unique incoming neighbors:")
print(unique_in_neighbors_full.describe())

print("\nUnique outgoing neighbors:")
print(unique_out_neighbors_full.describe())
```

```python
node_features["unique_in_neighbors"] = unique_in_neighbors_full
node_features["unique_out_neighbors"] = unique_out_neighbors_full

print(node_features.shape)

display(
    node_features[
        [
            "txId",
            "in_degree",
            "out_degree",
            "total_degree",
            "fan_in_out_ratio",
            "unique_in_neighbors",
            "unique_out_neighbors"
        ]
    ].head(10)
)
```

```python
node_features["neighbor_count"] = (
    node_features["unique_in_neighbors"] +
    node_features["unique_out_neighbors"]
)

print(node_features.shape)

display(
    node_features[
        [
            "txId",
            "unique_in_neighbors",
            "unique_out_neighbors",
            "neighbor_count"
        ]
    ].head(10)
)
```

```python
node_features["flow_balance"] = (
    (node_features["in_degree"] - node_features["out_degree"]) /
    node_features["total_degree"]
)

print(node_features.shape)

display(
    node_features[
        [
            "txId",
            "in_degree",
            "out_degree",
            "total_degree",
            "fan_in_out_ratio",
            "flow_balance"
        ]
    ].head(10)
)
```

```python
time_transaction_count = (
    node_features.groupby("Time_step")["txId"]
    .transform("count")
)

node_features["timestep_transaction_count"] = time_transaction_count

print(node_features.shape)

display(
    node_features[
        ["txId", "Time_step", "timestep_transaction_count"]
    ].head(10)
)
```

```python
min_time = node_features["Time_step"].min()
max_time = node_features["Time_step"].max()

node_features["time_position"] = (
    (node_features["Time_step"] - min_time) /
    (max_time - min_time)
)

print(node_features.shape)

display(
    node_features[
        ["txId", "Time_step", "time_position"]
    ].head(10)
)
```

```python
node_features["in_degree_ratio"] = (
    node_features["in_degree"] /
    node_features["total_degree"]
)

node_features["out_degree_ratio"] = (
    node_features["out_degree"] /
    node_features["total_degree"]
)

print(node_features.shape)

display(
    node_features[
        [
            "txId",
            "in_degree",
            "out_degree",
            "total_degree",
            "in_degree_ratio",
            "out_degree_ratio"
        ]
    ].head(10)
)
```

```python
import numpy as np

node_features["log_in_degree"] = np.log1p(node_features["in_degree"])
node_features["log_out_degree"] = np.log1p(node_features["out_degree"])
node_features["log_total_degree"] = np.log1p(node_features["total_degree"])
node_features["log_neighbor_count"] = np.log1p(node_features["neighbor_count"])

print(node_features.shape)

display(
    node_features[
        [
            "txId",
            "in_degree",
            "out_degree",
            "total_degree",
            "neighbor_count",
            "log_in_degree",
            "log_out_degree",
            "log_total_degree",
            "log_neighbor_count"
        ]
    ].head(10)
)
```

```python
time_degree_mean = (
    node_features.groupby("Time_step")["total_degree"]
    .transform("mean")
)

node_features["degree_vs_time_mean"] = (
    node_features["total_degree"] /
    time_degree_mean.replace(0, np.nan)
).fillna(0)

print(node_features.shape)

display(
    node_features[
        [
            "txId",
            "Time_step",
            "total_degree",
            "timestep_transaction_count",
            "degree_vs_time_mean"
        ]
    ].head(10)
)
```

```python
node_features["high_in_degree"] = (
    node_features["in_degree"] >= 10
).astype(int)

node_features["high_out_degree"] = (
    node_features["out_degree"] >= 10
).astype(int)

node_features["high_total_degree"] = (
    node_features["total_degree"] >= 10
).astype(int)

print(node_features.shape)

display(
    node_features[
        [
            "txId",
            "in_degree",
            "out_degree",
            "total_degree",
            "high_in_degree",
            "high_out_degree",
            "high_total_degree"
        ]
    ].head(10)
)
```

```python
node_features["flow_imbalance"] = (
    (node_features["in_degree"] - node_features["out_degree"]).abs()
    / node_features["total_degree"]
)

print(node_features.shape)

display(
    node_features[
        [
            "txId",
            "in_degree",
            "out_degree",
            "total_degree",
            "flow_balance",
            "flow_imbalance"
        ]
    ].head(10)
)
```

```python
original_feature_cols = [
    col for col in features.columns
    if col.startswith("feature_")
]

node_features["feature_mean"] = features[original_feature_cols].mean(axis=1)
node_features["feature_std"] = features[original_feature_cols].std(axis=1)
node_features["feature_min"] = features[original_feature_cols].min(axis=1)
node_features["feature_max"] = features[original_feature_cols].max(axis=1)
node_features["feature_median"] = features[original_feature_cols].median(axis=1)

print(node_features.shape)

display(
    node_features[
        [
            "txId",
            "feature_mean",
            "feature_std",
            "feature_min",
            "feature_max",
            "feature_median"
        ]
    ].head(10)
)
```

```python
node_features["extreme_feature_count_2"] = (
    features[original_feature_cols].abs() > 2
).sum(axis=1)

node_features["extreme_feature_count_3"] = (
    features[original_feature_cols].abs() > 3
).sum(axis=1)

node_features["extreme_feature_count_5"] = (
    features[original_feature_cols].abs() > 5
).sum(axis=1)

print(node_features.shape)

display(
    node_features[
        [
            "txId",
            "extreme_feature_count_2",
            "extreme_feature_count_3",
            "extreme_feature_count_5"
        ]
    ].head(10)
)
```
